```python
import os
import re
import pathlib as pl
```

```python
os.chdir('..')
home = pl.Path(os.getcwd())
print('home is at: ',home)

from src.hdf import *

inputs = home/'inputs'
outputs_base = home/'outputs'

#user to set variables for the project. tagged as parameter for papermill runs
project = 'wy_fy22'
export_folder = outputs_base/project/'trial_us_to_ds_events'
```

```python
import json
#export master combined list to json. should only be done once cannot be redone or may overwrite previous list.
with open(export_folder/'completed_event_dictionary.json', "r") as outfile2:
    event_dict = json.load(outfile2)
with open(export_folder/'completed_event_dictionary.json', "r") as outfile2:
    dss_links = json.load(outfile2)
```

```python
copy_dss_store = {}
missing_dss = {}
```

```python
for huc in event_dict.keys():
    print(huc)
    copy_events = {}
    missing_dss[huc] = {}
    for ri in event_dict[huc].keys():
        #print(ri)
        events_list = []
        link_dict = {}
        
        for event in event_dict[huc][ri]:
            #print(event)
            #check in hydrology folder

            out_fold = f'wy_gdg_{huc[0:8]}*'
            in_fold = f'HUC{huc[0:8]}/{huc}'
            avail_dss = glob.glob(str(outputs_base/project/out_fold/'Hydrology'/'*.dss'))
            #already_collected_extra = glob.glob(str(inputs/project/'us_dss'/in_fold/'*'/'*.dss'))
            #avail_dss = events#+already_collected_extra
            re_event = event.replace('-','[-_]')
            re_event = re_event.replace('R','[\s\S]+R-*')
            r = re.compile(re_event+'_output.dss')
            links = list(filter(r.match,avail_dss))
            #print(links)
            if links:
                in_huc = 0
                other_huc = 0
                for link in links:
                    if link.find(huc) < 0:
                        other_huc=1
                    else:
                        in_huc=1
                if in_huc == 0 and other_huc == 1:
                    copy_events[event] = link
                link_dict[event] = link
            else:
                events_list.append(event)
                link_dict[event] = []
        #print(huc,ri,link_dict)
        dss_links[huc][ri] = link_dict
        missing_dss[huc][ri] =  events_list
    copy_dss_store[huc] = copy_events
```

```python
for huc in copy_dss_store.keys():
    for event, link in copy_dss_store[huc].items():
        print(link, f'is being copied into huc {huc} hydrology folder') 
        os.makedirs(inputs/project/'transfer_dss'/huc[:8],exist_ok = True)
        shutil.copy(link,inputs/project/'transfer_dss'/huc[:8])
```
